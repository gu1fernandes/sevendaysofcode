```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Configura o visual padrão dos gráficos
sns.set_theme(style="whitegrid")

# Substitua pela URL Raw do GitHub ou o caminho do arquivo baixado (ex: 'spotify_data.csv')
df = pd.read_csv('dataset.csv')

# Visualizar as 5 primeiras linhas
display(df.head())
```

```python
# Dimensão da base (linhas e colunas) e tipos de dados
print(f"O dataset possui {df.shape[0]} linhas e {df.shape[1]} colunas.\n")
df.info()

# Contagem de valores nulos e duplicados
print("\n--- Valores Nulos por Coluna ---")
print(df.isnull().sum())

print(f"\nTotal de linhas duplicadas: {df.duplicated().sum()}")

# Estatísticas descritivas básicas (média, min, max, quartis)
display(df.describe())
```

```python
# 1. As 10 músicas mais populares
top_musicas = df[['track_name', 'artists', 'popularity']].sort_values(by='popularity', ascending=False).head(10)
print("Top 10 Músicas Mais Populares:")
display(top_musicas)

# 2. Os 10 artistas com maior popularidade média (considerando artistas com mais de 5 músicas)
artistas_populares = df.groupby('artists').filter(lambda x: len(x) > 5)
top_artistas = artistas_populares.groupby('artists')['popularity'].mean().sort_values(ascending=False).head(10)

plt.figure(figsize=(10, 5))
sns.barplot(x=top_artistas.values, y=top_artistas.index, palette="mako")
plt.title("Top 10 Artistas Mais Populares (Média)")
plt.xlabel("Popularidade")
plt.ylabel("Artista")
plt.show()
```

```python
# Mapa de calor (Heatmap) de correlação entre as variáveis numéricas
colunas_numericas = ['popularity', 'danceability', 'energy', 'loudness', 'speechiness', 'acousticness', 'instrumentalness', 'liveness', 'valence', 'tempo']

plt.figure(figsize=(12, 8))
matriz_correlacao = df[colunas_numericas].corr()
sns.heatmap(matriz_correlacao, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
plt.title("Correlação entre Características Musicais e Popularidade")
plt.show()
```
